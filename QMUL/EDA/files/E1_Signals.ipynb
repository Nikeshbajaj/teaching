## Elementary Signals

### Unit step

Unit step signal is signal, which has value 1 for t>0 and 0 otherwise. For continues time signal, at t=0, value is 0.5, however for discreet time, it is 1.


\begin{equation}
  u(t)=\begin{cases}
    1, & \text{if $t>0$}.\\
    0, & \text{otherwise}.
  \end{cases}
\end{equation}

```python
fs = 100         #Sampling Frquency
#Ts = 1/fs       #Sampling duration, -  duration between two consequent samples


t1, t2 = -5, 5   #from -5 to 5 sec

#t = np.arange(t1,t2+Ts,Ts)

t = np.linspace(t1,t2,fs*(t2-t1))

ut = 1*(t>0)

plt.figure()
plt.plot(t,ut)
plt.xlabel('time (s)')
plt.ylabel('u(t)')
plt.title('Unit Step signal: u(t)')
plt.show()
```

### Ramp function

\begin{equation}
  r(t)=\begin{cases}
    t, & \text{if $t>0$}.\\
    0, & \text{otherwise}.
  \end{cases}
\end{equation}

```python
fs = 100        #Sampling Frquency
t1, t2 = -5, 5   #from -5 to 5 sec

#Ts = 1/fs
#t = np.arange(t1,t2+Ts,Ts)

t = np.linspace(t1,t2,fs*(t2-t1))

rt = t.copy()
rt[rt<0]=0

plt.figure()
plt.plot(t,rt)
plt.xlabel('time (s)')
plt.ylabel('r(t)')
plt.title('Ramp signal: r(t)')
plt.show()
```

### Gate function

\begin{equation}
  g(t)=\begin{cases}
    1, & \text{if $|t|<T$}.\\
    0, & \text{otherwise}.
  \end{cases}
\end{equation}

```python
fs = 100        #Sampling Frquency
t1, t2 = -5, 5   #from -5 to 5 sec


#Ts = 1/fs
#t = np.arange(t1,t2+Ts,Ts)
t = np.linspace(t1,t2,fs*(t2-t1))

T = 0.5
gt = np.zeros(len(t))
gt[np.abs(t)<T]=1

plt.figure()
plt.plot(t,gt)
plt.xlabel('time (s)')
plt.ylabel('g(t)')
plt.title('Gate function: g(t)')
plt.show()
```

### Signum function

\begin{equation}
  s(t)=\begin{cases}
    1, & \text{if $t>0$}.\\
    0, & \text{if $t=0$}.\\
    -1, & \text{if $t<0$}.
  \end{cases}
\end{equation}

```python
fs = 100        #Sampling Frquency
t1, t2 = -5, 5   #from -5 to 5 sec

#Ts = 1/fs
#t = np.arange(t1,t2+Ts,Ts)
t = np.linspace(t1,t2,fs*(t2-t1))

st = np.sign(t)

plt.figure()
plt.plot(t,st)
plt.xlabel('time (s)')
plt.ylabel('s(t)')
plt.show()
```

### Sinc function

\begin{equation}
  x(t)= \frac{sin(\pi t)}{\pi t}
\end{equation}

```python
fs = 100         #Sampling Frquency
t1, t2 = -5, 5   #from -5 to 5 sec

#Ts = 1/fs
#t = np.arange(t1,t2+sd,sd)
t = np.linspace(t1,t2,fs*(t2-t1))

xt = np.sin(np.pi*t)/(np.pi*t)
xt[t==0]=1


plt.figure()
plt.plot(t,xt)
plt.xlabel('time (s)')
plt.ylabel('s(t)')
plt.show()
```

### Exponatial function

\begin{equation}
  x(t)= exp(-at)
\end{equation}


```python
fs = 100         #Sampling Frquency
t1, t2 = -5, 5   #from -5 to 5 sec


t = np.linspace(t1,t2,fs*(t2-t1))


a = 0.5
x1 = np.exp(-a*t)

a = -0.5
x2 = np.exp(-a*t)


plt.figure()
plt.plot(t,x1, label=r'$exp(-0.5t)$')
plt.plot(t,x2, label=r'$exp(0.5t)$')
plt.xlabel('time (s)')
plt.ylabel('x(t)')
plt.legend()
plt.show()
```

### Sinusodal function

$$x(t) = Sin(2 \pi f t)$$

### Genaralised Exponential function

\begin{equation}
  x(t) = Ce^{\alpha t}
\end{equation}

* $C$ can be real or complex
* $\alpha$ cab be real or complex


Cases:

* $C$, $\alpha$ both real, $\alpha$>0

    - $x(t) = 3e^{5t}$
    - increasing explonential
    
* $C$, $\alpha$ both real, $\alpha$<0

    - $x(t) = 3e^{-5t}$
    - decreasing explonential
    
* $C$, real and  $\alpha$ complex, pure imaginary

    - $x(t) = 3e^{j5t}$
    - $3e^{j5t} = cos(5t) + jsin(5t)$
    - sinusoidal
    
* $C$, real and  $\alpha$ complex

    - $x(t) = 3e^{ (2+ j5) t}$
    - $x(t) = 3e^{2t} e^{j5t}$

### Complex Exponential ~ Sinusodal



\begin{equation}
  x(t) = e^{j2\pi ft} =  exp(j2\pi ft)  = cos(2\pi ft) + jsin(2\pi ft)
\end{equation}

```python
fs = 100        #Sampling Frquency

f = 1           #Hz
t1, t2 = -3, 3   #from 0 to 3 sec

#Ts = 1/fs
#t  = np.arange(t1,t2+Ts,Ts)

t  = np.linspace(t1,t2,fs*(t2-t1))
x1 = np.exp(1j*2*np.pi*f*t)

plt.figure(figsize=(10,2))
plt.plot(t,x1.real,label='real')
plt.plot(t,x1.imag,label='imaginary')
plt.xlabel('time (s)')
plt.ylabel('amplitude')
plt.legend(loc=1)
plt.show()
```

## Operations on signals

### Addition

$$y(t) = x_1(t) + x_2(t)$$


```python
# Sum of Sins

fs = 100        #Sampling Frquency
t1, t2 = -3, 3   #from -3 to 3 sec
t = np.linspace(t1,t2,fs*(t2-t1))


x1 = np.sin(2*np.pi*1*t)
x2 = np.sin(2*np.pi*3*t)
x3 = np.sin(2*np.pi*4*t)

#y = x1 + x2 + x3
y = 1*x1 + 0.5*x2 + 0.2*x3 

plt.figure(figsize=(10,3))
plt.plot(t,y)
plt.xlabel('time (s)')
plt.ylabel('Amplitude')
plt.show()
```

### Time Reversal

$$y(t) = x(-t)$$

```python

fs = 100        #Sampling Frquency
t1, t2 = -3, 5   #from -5 to 5 sec

#Ts = 1/fs
#t = np.arange(t1,t2+Ts,Ts)

t = np.linspace(t1,t2,fs*(t2-t1))

x = t.copy()
x[x<0]=0


y  = x[::-1]
t1 = -t[::-1]

plt.figure()
plt.plot(t,x,label='x(t)')
plt.plot(t1,y,label='y(t) = x(-t)')
plt.xlabel('time (s)')
plt.ylabel('r(t)')
plt.title('Ramp signal: r(t)')
plt.legend()
plt.show()
```

## Creating other signals from sinusodal

### Square signal


$$x(t) = sgn(sin(2 \pi ft))$$

```python
fs = 100        #Sampling Frquency

f = 1           #Hz
t1, t2 = -3, 3   #from 0 to 3 sec

#Ts = 1/fs
#t = np.arange(t1,t2+Ts,Ts)

t = np.linspace(t1,t2, fs*(t2-t1))
x1 = np.sin(2*np.pi*f*t)
x2 = np.sign(x1)

plt.figure(figsize=(10,2))
plt.plot(t,x1,label=r'$x_1(t)$: sin')
plt.plot(t,x2,label=r'$x_2(t)$: square')
plt.xlabel('time (s)')
plt.ylabel('amplitude')
plt.legend(loc=1)
plt.show()
```

### Square signal with different width

### Triangular

$$x(t) = \frac{2}{\pi}sin^{-1}(sin(2 \pi ft))$$

```python
fs = 100        #Sampling Frquency
f = 1           #Hz
t1, t2 = -3, 3   #from 0 to 3 sec
t = np.linspace(t1,t2, fs*(t2-t1))
x1 = np.sin(2*np.pi*f*t)


x4 = (2/np.pi)*np.arcsin(x1)
plt.figure(figsize=(10,2))
plt.plot(t,x4)
#plt.plot(t,np.gradient(x4)*10)
plt.xlabel('time (s)')
plt.ylabel('amplitude')
plt.show()
```

### Sawtooth


$$x(t) = \frac{1}{\pi}tan^{-1}(sin(2 \pi ft), cos(2 \pi ft))$$

```python
fs = 100        #Sampling Frquency
f = 1           #Hz
t1, t2 = -3, 3   #from 0 to 3 sec
t = np.linspace(t1,t2, fs*(t2-t1))

x1 = np.sin(2*np.pi*f*t)
x2 = np.cos(2*np.pi*f*t)
x4 = (1/np.pi)*np.arctan2(x1,x2)
plt.figure(figsize=(10,2))
plt.plot(t,x4)
#plt.plot(t,np.gradient(x4)*1)
plt.xlabel('time (s)')
plt.ylabel('amplitude')
plt.show()
```
