# Lab 2: Electroenceplogram -  a walk through 

Electroenceplography (EEG) records the tiny electrical activities on the scalp. These electrical activities are the summations of millions of nuerons' (Brain Cells) potential, generated within the cortex.


EEG records these potential voltages at various locations (sites) on the scalp, which are varying over the time.

An electrode place on scalp, cover a bigger cortext region, therefore, EEG could not record nueronal activity of region.

The signal travels to scalp (thick skull bones) from cortex is very small (micro-volt), therefore it is contaminated by various types of noise and unwanted intereferences. While amplifying, these noise amplified too.

Electrodes are small sensors, placed on the scalp, sometimes with conductive gel to record the signal. Their placement on the scalp is based on International 10-20 System. Each electrod location is named as charactor(s) followed by a number (e.g. 'F2'). The charactors are named after the region of brain, such as Prefrontal (Fp), Frontal (F), Occupital (O), Parietal (P), Temporal (T), and center (C).

## Lollypop plot

Instead of a line plot (above plots), let's plot each sample seperetely to see how number of samples are important

```python
plt.figure(figsize=(15,3))
plt.stem(t,x)
plt.plot(t,x,'--',alpha=0.5)
plt.xlim([t[0],t[-1]])
plt.xlabel('time (s)')
plt.ylabel('amplitude (µV)')
plt.title('Electrode : '+ch_name)
plt.grid()
plt.show()
```

**Downsampling**

Let's try to downsample this one second segement. Downsampling with factor *D* means to select every other *D*th samples. This results in low sampling rate. For examples, if a signal is sampled at 100Hz (100 samples per second), then downsampling with *D=2*, will results same signal as sampled at 100/D  = 100/2 = 50Hz.

With numpy array, downsampling is very straightforward by using ::D argument.

```python
D = 2
x1 = x[::D]
t1 = t[::D]

plt.figure(figsize=(15,3))
plt.stem(t1,x1)
plt.plot(t1,x1,'--',alpha=0.5)
plt.xlim([t[0],t[-1]])
plt.xlabel('time (s)')
plt.ylabel('amplitude (µV)')
plt.title('Electrode : '+ch_name)
plt.grid()
plt.show()
```

# Formats of EEG Files

There are various devices to record EEG data and these devices uses different file and data structures to record the signals. While exporting the data, they use different file formats. Commonly used file formate for EEG is 'EDF' (European Data Format - https://en.wikipedia.org/wiki/European_Data_Format), however some devices allow you to extract the data in CSV formate. Regardless of device supporting file format, many researchers and experimentors share their data in many different formats to make it easy to use particular analysis libraries.

In this section, we are going to see how to read different files and extract signals and the relevent information from files.


* CSV (Comma Seperated Values)
* EDF (European Data Format)
* MAT (MATLAB)
* NPY (Numpy)

## EDF Files

**EDF File:** EDF is European Document Format, which allows to store data in a structured way. It is widely used for electronics devices and sensor data in many different fields, such as medical equipments. It allows to records the sensor information, recording details, and specifications regards to each channels and experiment.


We will use four EDF files, we will check them one-by-one. To read EDF file, we will use, `sp.io.read_bdf`, which extracts the signals, sampling frequency and channel labels and all the information.
